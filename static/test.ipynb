```python
from pydantic import BaseModel

class PydanticUser(BaseModel):
    name:str

user2 = PydanticUser(name="홍길동")  # 정상 작동! 
print("Pydantic 결과:", user2.name)
print("Pydantic 데이터 타입:", type(user2.name))
user2.name, type(user2.name)
```

```python
class UserRegister(BaseModel):
    user_id: int 
    username: str         
    email: str
    is_admin: bool = False # 선택 입력, 기본값은 False
```

```python
user=UserRegister(
    user_id =1,   
    username= "홍길동",         
    email= "gong@example.com"
)

print(user)
```

```python
input_data = {
    "user_id": "1004",
    "username": "codemaster",
    "email": "code@example.com"
}
```

```python
user = UserRegister(**input_data)
print("생성된 객체:", user)
print("유저 ID 타입:", type(user.user_id))

user_dict = user.model_dump()
user_json = user.model_dump_json()

print("딕셔너리 변환:", user_dict)
print("JSON 문자열 변환:", user_json)
```
